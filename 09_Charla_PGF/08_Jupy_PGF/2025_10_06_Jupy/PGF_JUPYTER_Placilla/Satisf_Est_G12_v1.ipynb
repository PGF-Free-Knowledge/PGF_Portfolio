```python
# ============================================
# Resultados de Satisfacción Estudiantil
# Gráficos para dos cursos paralelos
# Autor: Paul Gálvez / Grupo 12
# ============================================

import pandas as pd
import matplotlib.pyplot as plt

# -----------------------------
# Datos base (dos cursos)
# -----------------------------
data = {
    "Categoría de Satisfacción": [
        "Muy satisfactoria",
        "Satisfactoria",
        "Medianamente satisfactoria",
        "Poco satisfactoria"
    ],
    "Curso A (%)": [50, 30, 15, 5],
    "Curso B (%)": [55, 25, 15, 5]
}

df = pd.DataFrame(data)

# -----------------------------
# Gráfico 1: Barras comparativo
# -----------------------------
plt.figure(figsize=(8,5))
bar_width = 0.35
x = range(len(df))

plt.bar(x, df["Curso A (%)"], width=bar_width, label="Curso A", color="#2E8B57")
plt.bar([p + bar_width for p in x], df["Curso B (%)"], width=bar_width, label="Curso B", color="#A9A9A9")

plt.xlabel("Categoría de satisfacción")
plt.ylabel("Porcentaje (%)")
plt.title("Comparativo de satisfacción estudiantil por curso (n = 30 cada uno)")
plt.xticks([p + bar_width/2 for p in x], df["Categoría de Satisfacción"], rotation=15)
plt.legend()
plt.tight_layout()
plt.savefig("grafico_comparativo.png", dpi=300)
plt.show()

# -----------------------------
# Gráfico 2: Circulares por curso
# -----------------------------
fig, axes = plt.subplots(1, 2, figsize=(10,5))
colors = ["#2E8B57", "#90EE90", "#FFD700", "#DC143C"]

axes[0].pie(df["Curso A (%)"], labels=df["Categoría de Satisfacción"],
            autopct='%1.0f%%', colors=colors, textprops={'fontsize': 9})
axes[0].set_title("Distribución de satisfacción – Curso A")

axes[1].pie(df["Curso B (%)"], labels=df["Categoría de Satisfacción"],
            autopct='%1.0f%%', colors=colors, textprops={'fontsize': 9})
axes[1].set_title("Distribución de satisfacción – Curso B")

plt.tight_layout()
plt.savefig("graficos_circulares.png", dpi=300)
plt.show()

print("\n✅ Gráficos generados y guardados como:")
print("   - grafico_comparativo.png")
print("   - graficos_circulares.png")
```
